# Mox Chatbot User Research Analysis

This notebook is a lightweight wrapper around `src/run_analysis.py`. It runs demo mode by default using synthetic data only.

```python
from pathlib import Path
import sys

def find_repo_root(start):
    current = Path(start).resolve()
    for candidate in [current, *current.parents]:
        if (candidate / "src" / "run_analysis.py").exists():
            return candidate
    raise RuntimeError("Could not find repository root containing src/run_analysis.py")

repo_root = find_repo_root(Path.cwd())
sys.path.insert(0, str(repo_root))
from src.run_analysis import run_analysis
```

## Demo run

Demo mode uses synthetic aggregate variables. It does not access the private questionnaire.

```python
run_analysis(
    "demo",
    repo_root / "data" / "synthetic_example.csv",
    repo_root / "local_results" / "demo",
)
```

```python
import pandas as pd

coef_table = pd.read_csv(repo_root / "local_results" / "demo" / "tables" / "model_coefficients.csv")
display_columns = ["model", "term", "n", "coef", "robust_p_hc3", "robust_ci95_hc3"]
coef_table.loc[coef_table["term"] != "const", display_columns]
```

## Optional canonical run

Canonical mode requires a private questionnaire path supplied locally. Keep `RUN_CANONICAL = False` for public use.

```python
RUN_CANONICAL = False
PRIVATE_QUESTIONNAIRE = None

if RUN_CANONICAL:
    if PRIVATE_QUESTIONNAIRE is None:
        raise ValueError("Set PRIVATE_QUESTIONNAIRE to a local private xlsx path before running canonical mode.")
    run_analysis(
        "canonical",
        PRIVATE_QUESTIONNAIRE,
        repo_root / "local_results" / "canonical",
    )
```
