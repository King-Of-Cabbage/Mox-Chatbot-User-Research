Run demo mode by default. Canonical mode is disabled unless `RUN_CANONICAL` is changed locally.


```python
from pathlib import Path
import sys

def find_repo_root(start):
    current = Path(start).resolve()
    for candidate in [current, *current.parents]:
        if (candidate / 'src' / 'run_analysis.py').exists():
            return candidate
    raise RuntimeError('Could not find repository root containing src/run_analysis.py')

repo_root = find_repo_root(Path.cwd())
sys.path.insert(0, str(repo_root / 'src'))
from run_analysis import run_analysis
```

```python
run_analysis(
    "demo",
    repo_root / "data" / "synthetic_example.csv",
    repo_root / "local_results" / "demo"
)
```
